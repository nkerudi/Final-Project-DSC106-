```python
catalog_df = pd.read_csv('https://storage.googleapis.com/cmip6/cmip6-zarr-consolidated-stores.csv')
```

```python
gcs = gcsfs.GCSFileSystem(token='anon')
```

```python
def get_catalog(id):
    zstore_id = catalog_df.loc[id].zstore
    mapper = gcs.get_mapper(zstore_id)
    ds_id = xr.open_zarr(mapper, consolidated=True)
    return ds_id

def verify_lat_lon_time(df):
    if 'lat' not in df.coords:
        return "Latitude column is missing"
    if 'lon' not in df.coords:
        return "Longitude column is missing"
    if 'time' not in df.coords:
        return "Time column is missing"
    return "Latitude, Longitude, and Time columns are present"
```

```python
def calculate_anom(ds, var, coarsen_var=2, year_agg=5):
    var_df = ds[var]
    
    # detect grid type
    spatial_dims = [d for d in var_df.dims if d not in ('time',)]
    # build coarsen kwargs dynamically based on actual dim names
    coarsen_kwargs = {dim: coarsen_var for dim in spatial_dims}
    
    var_df = var_df.coarsen(coarsen_kwargs, boundary="trim").mean()
    yearly = var_df.resample(time=f"{year_agg}Y").mean()
    baseline = yearly.isel(time=0)
    anom = yearly - baseline
    return anom
```

```python
def plot_ssp_map(cmip6_data, ssp, var, var_name, vmax=None):
    data = cmip6_data.load()
    if data.lat.ndim == 2:
        lats = data.lat.values[:, 0]  # take first column if lat is 2D
        lons = data.lon.values[0, :]  # take first row if lon is 2D
    else:
        lats = data.lat.values
        lons = data.lon.values
    
    years = data.time.dt.year.values
    
    if vmax is None:
        vmax = float(abs(data.max().values))

    def make_heatmap(i):
        z = data.isel(time=i).values
        return go.Heatmap(
            z=z,
            x=lons,
            y=lats,
            colorscale='Balance',
            zmin=-vmax,
            zmax=vmax,
            colorbar=dict(title=var_name),
            hovertemplate="Lat: %{y}<br>Lon: %{x}<br>Value: %{z}<extra></extra>",
        )

    frames = [
        go.Frame(
            data=[make_heatmap(i)],
            name=str(years[i]),
            traces=[0]
        ) for i in range(len(years))
    ]

    sliders = [{
        "steps": [
            {
                "args": [[f.name], {"frame": {"duration": 0, "redraw": True}, "mode": "immediate"}],
                "label": f.name,
                "method": "animate",
            } for f in frames
        ],
        "active": 0,
        "currentvalue": {"prefix": "Year: "},
        "pad": {"t": 50}
    }]

    fig = go.Figure(data=[make_heatmap(0)], frames=frames)

    fig.update_layout(
        title=f"{var_name} Evolution ({ssp})",
        width=900,
        height=600,
        sliders=sliders,
        plot_bgcolor="gray",  # shows through NaN cells = ocean
        xaxis=dict(range=[lons.min(), lons.max()], showgrid=False, zeroline=False),
        yaxis=dict(range=[lats.min(), lats.max()], showgrid=False, zeroline=False),
    )

    fig.show()
```

```python
def da_to_dict(da):
    # handle both 1D (atmospheric) and 2D (ocean curvilinear) grids
    if da.lat.ndim == 2:
        # curvilinear: flatten 2D lat/lon
        lats_2d = da.lat.values
        lons_2d = da.lon.values
        
        # normalize lons
        lons_2d = np.where(lons_2d > 180, lons_2d - 360, lons_2d)
        
        lats = lats_2d.flatten().tolist()
        lons = lons_2d.flatten().tolist()
        nLat = da.sizes['y']
        nLon = da.sizes['x']
        is_curvilinear = True
    else:
        # regular grid: 1D lat/lon
        lats = da.lat.values.tolist()
        lons_raw = da.lon.values
        lons = [float(l - 360 if l > 180 else l) for l in lons_raw]
        nLat = len(lats)
        nLon = len(lons)
        is_curvilinear = False

    years = da.time.dt.year.values.tolist()

    frames = []
    for i in range(len(da.time)):
        flat = da.isel(time=i).values.flatten()
        frames.append([None if np.isnan(v) else round(float(v), 5) for v in flat])

    return {
        "years": years,
        "lats": lats,
        "lons": lons,
        "nLat": nLat,
        "nLon": nLon,
        "frames": frames,
        "curvilinear": is_curvilinear  # flag so JS knows the grid type
    }

def export_precomputed(data_dict, filename):
    print("Writing to disk...")
    with open(filename, "w") as f:
        json.dump(data_dict, f, separators=(',', ':'))
    size_mb = round(os.path.getsize(filename) / 1e6, 1)
    print(f"Done — {filename} ({size_mb} MB)")
    return size_mb
```

```python
# export temperature anomaly data
tos_anom_ssp126_dict = da_to_dict(calculate_anom(get_catalog(tos_ds['ssp126']), var='tos'))
tos_anom_ssp245_dict = da_to_dict(calculate_anom(get_catalog(tos_ds['ssp245']), var='tos'))
tos_anom_ssp370_dict = da_to_dict(calculate_anom(get_catalog(tos_ds['ssp370']), var='tos'))
tos_anom_ssp585_dict = da_to_dict(calculate_anom(get_catalog(tos_ds['ssp585']), var='tos'))
```

```python
for ssp, data_dict in zip(['ssp126', 'ssp245', 'ssp370', 'ssp585'], 
                          [tos_anom_ssp126_dict, tos_anom_ssp245_dict, tos_anom_ssp370_dict, tos_anom_ssp585_dict]):
    export_precomputed(data_dict, f"tos_anom_{ssp}.json")
```

```python
# export global temperature anomaly data
ts_anom_ssp126_dict = da_to_dict(calculate_anom(get_catalog(ts_ds['ssp126']), var='ts'))
ts_anom_ssp245_dict = da_to_dict(calculate_anom(get_catalog(ts_ds['ssp245']), var='ts'))
ts_anom_ssp370_dict = da_to_dict(calculate_anom(get_catalog(ts_ds['ssp370']), var='ts'))
ts_anom_ssp585_dict = da_to_dict(calculate_anom(get_catalog(ts_ds['ssp585']), var='ts'))

for ssp, data_dict in zip(['ssp126', 'ssp245', 'ssp370', 'ssp585'], 
                          [ts_anom_ssp126_dict, ts_anom_ssp245_dict, ts_anom_ssp370_dict, ts_anom_ssp585_dict]):
    export_precomputed(data_dict, f"ts_anom_{ssp}.json")
```
